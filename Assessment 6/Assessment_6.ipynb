```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder \
    .appName("Week6 Assignment") \
    .getOrCreate()
```

```python
df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("data/source.csv")

print("Original Data")

df.show()
```

```python
print("Schema")

df.printSchema()
```

```python
print("Select Product ID and Price")

df.select("product_id", "price").show()
```

```python
#Filter Electronics
print("Electronics Products")

df.filter(df.category=="Electronics").show()
```

```python
#Rename Column
df = df.withColumnRenamed("old_name","new_name")

print("Renamed Column")

df.show()
```

```python
#Cast Price
df = df.withColumn(
    "price",
    col("price").cast("double")
)

print("Price Converted")

df.printSchema()
```

```python
#Add New Column
df = df.withColumn(
    "final_price",
    col("base_price")*1.18
)

print("Final Price")

df.show()
```

```python
#Filter Orders
print("Completed Orders")

df.filter(
    (col("status")=="Completed") &
    (col("amount")>1000)
).show()
```

```python
#Handle Null Values
print("Non Null User IDs")

df.filter(
    col("user_id").isNotNull()
).show()
```

```python
#Filter OR Condition
print("North OR High")

df.filter(
    (col("region")=="North") |
    (col("priority")=="High")
).show()
```

```python
#Wide Transformation
print("Group By Category")

df.groupBy("category").count().show()
```

```python
#Save as CSV
df.write \
.mode("overwrite") \
.option("header",True) \
.csv("output/csv_output")
```

```python
#Save as Parquet
df.write \
.mode("overwrite") \
.parquet("output/parquet_output")
```

```python
#Read Parquet
parquet_df = spark.read.parquet("output/parquet_output")

print("Reading Parquet")

parquet_df.show()
```

```python
#Stop Spark
spark.stop()
```
